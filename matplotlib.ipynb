# BAR PLOT

```python
import matplotlib.pyplot as plt
x = ["python", "C" , "C++" , "Java" , "JavaScript"]
y = [85 , 70 , 65 , 82 , 78]
plt.xlabel("language")
plt.ylabel("number")
plt.title("CSE")
plt.bar(x , y , width = 0.5 , color = "pink" , edgecolor = "white")
plt.show()
```

# SCATTER PLOT

```python
import matplotlib.pyplot as plt
x = [ 1 , 2 , 3 , 5, 6 , 7]
y = [2 , 3, 7, 8, 4, 9]
colors = ["red" , "yellow" , "green", "blue" , "red", "yellow"]
plt.xlabel("Day")
sizes = [400 , 200 , 100 , 300 , 200 , 300]
plt.ylabel("Number")
plt.title("Scatter Plot" , fontsize = 20)
plt.scatter(x,y , c = colors , s = sizes)
plt.show()
```

# PIE CHART

```python
import matplotlib.pyplot as plt
x = [10 , 20 , 30 , 40]
y = ["C" , "C++" , "Java" , "JavaScript"]
ex = [0.4 , 0.0 , 0.0 , 0.0]
plt.pie(x , labels = y , explode = ex , autopct = "%0.1f%%"  )
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10 , 20 , 30 , 40]
x1 = [40 , 30 , 20 , 10]
y = ["C" , "C++" , "Java" , "JavaScript"]

plt.pie(x , labels = y , radius = 1)
plt.pie(x1 , radius = 0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10 , 20 , 30 , 40]
x1 = [40 , 30 , 20 , 10]
y = ["C" , "C++" , "Java" , "JavaScript"]

c = ["red" , "yellow" , "green" , "blue" ]

plt.pie(x , labels = y ,  colors=c,
    wedgeprops={"width": 0.4})

plt.show()
```

# STEM PLOT

```python
import matplotlib.pyplot as plt
x = [1 , 2 , 3 , 4, 5, 6]
y = [2 , 2 , 5 , 7, 8 , 4]
plt.stem(x , y , linefmt = ":" , markerfmt = "ro" , bottom = 1 , basefmt = "g" , label = "python")
plt.legend()
plt.show()
```
